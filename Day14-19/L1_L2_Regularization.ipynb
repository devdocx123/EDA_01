```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import os
```

```python
folder_path = "ML files"
file_path = os.path.join(folder_path , 'Melbourne_housing_FULL.csv')
df = pd.read_csv(file_path)
df.head()
```

```python
df.columns
```

```python
df.nunique()
```

```python
df.shape
```

```python
col_to_use = ['Suburb', 'Rooms', 'Type', 'Method', 'SellerG',
        'Distance', 'Bedroom2', 'Bathroom', 'Car',
       'Landsize', 'BuildingArea', 'CouncilArea', 'Regionname', 'Propertycount' , 'Price']
df =df[col_to_use]
df.head()
```

```python
df.isna().sum()
```

```python
col_to_fill_zero = ['Distance' , 'Bedroom2' , 'Car','Bathroom' , 'Propertycount']
df[col_to_fill_zero] = df[col_to_fill_zero].fillna(0)
```

```python
df.isna().sum()
```

```python
df['Landsize'] = df['Landsize'].fillna(df['Landsize'].mean())
df['BuildingArea'] = df['BuildingArea'].fillna(df['BuildingArea'].mean())
```

```python
df.isna().sum()
```

```python
df.dropna(inplace = True)
df.isna().sum()
```

```python
df = pd.get_dummies(df , drop_first=True).astype(int)
df.head()
```

```python
df.shape
```

```python
X  = df.drop('Price' , axis = 1)
Y = df.Price
```

```python
X_train , X_test , Y_train , y_test = train_test_split(X ,Y ,test_size = 0.3 , random_state = 2)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
# Training model without L1 and L2 Regularization
model = LinearRegression()
model.fit(X_train , Y_train)
```

```python
model.score(X_test , y_test)
```

```python
model.score(X_train , Y_train) #look at the difference of acuracy between train and test splits.
```

```python
#Model Training using L1 regularization 
```

```python
from sklearn import linear_model


Lasso_reg = linear_model.Lasso(alpha=50 , max_iter = 100 , tol = 1.0)

Lasso_reg.fit(X_train , Y_train)
```

```python
Lasso_reg.score(X_train , Y_train)
```

```python
Lasso_reg.score(X_test , y_test)
```

```python
#Model Training using L2 regularization
from sklearn.linear_model import Ridge
ridge_reg = Ridge(alpha=50 , max_iter = 100 , tol = 1.0)
ridge_reg.fit(X_train , Y_train)
```

```python
ridge_reg.score(X_test ,y_test)
```

```python
ridge_reg.score(X_train ,Y_train)
```
