```python
from sklearn.model_selection import StratifiedKFold
skf = StratifiedKFold(n_splits = 3 , shuffle=True, random_state=42)
```

```python
# manual way of doing K_fold

score_lm = []
score_svc = []
score_rfc = []

def get_score(model , X_train, X_test , Y_train ,Y_test):
    model.fit(X_train , Y_train)
    return model.score(X_test , Y_test)
for train_index , test_index in skf.split(iris.data , iris.target):
    X_train , X_test = iris.data[train_index] , iris.data[test_index]
    Y_train ,Y_test  = iris.target[train_index] , iris.target[test_index]
    
    score_lm.append( get_score(LogisticRegression() , X_train , X_test , Y_train  ,Y_test))
   
    score_svc.append(get_score(SVC() , X_train , X_test , Y_train  ,Y_test))
       
    score_rfc.append(get_score(RandomForestClassifier(n_estimators = 40) , X_train , X_test , Y_train  ,Y_test))

```

```python
print("score_lm=", score_lm)
```

```python
print("score_svc =  " ,  score_svc)
```

```python
print("score_rfc =  " ,  score_rfc)
```

```python
# DOING SAME THING WITH THE HELP OF BUILT IN FUCTION THAT HELPS US IN AVOIDING THE WHOLE PARAGRAPH
from sklearn.model_selection import cross_val_score
```

```python
cross_val_score(LogisticRegression() , iris.data , iris.target ,cv=3)
```

```python
cross_val_score(SVC() , iris.data , iris.target ,cv=3)
```

```python
cross_val_score(RandomForestClassifier(n_estimators = 15) , iris.data , iris.target ,cv=3)
```

```python
# Parametric tuning > Used to pass diff parameters to different model to enhance auracy and make comparison on their basis
cross_val_score(SVC(C = 1 , kernel = 'linear') , iris.data , iris.target ,cv=3)
```

```python
cross_val_score(LogisticRegression(C = 5 , max_iter = 200) , iris.data , iris.target ,cv=3) # Improved
```

```python
cross_val_score(RandomForestClassifier(n_estimators = 10 , max_depth = 1) , iris.data , iris.target ,cv=3) #Decreased
```

```python
cross_val_score(RandomForestClassifier(n_estimators = 10 , max_depth = 1 , min_samples_split = 520) , iris.data , iris.target ,cv=3) #Decreased Also
```
